```python
import pandas as pd

# Load the healthy and corrupted files
healthy_file_path = '../final_article.csv'  # Replace with the path to the healthy file
corrupted_file_path = r"C:\Users\Parsa\Downloads\MML harvard files\final_articles.csv" # Replace with the path to the corrupted file

healthy_df = pd.read_csv(healthy_file_path)
corrupted_df = pd.read_csv(corrupted_file_path)

def fix_and_align_corrupted_file(healthy_df, corrupted_df):
    """
    Fix the corrupted file by aligning rows to match the structure of the healthy file.
    """
    fixed_rows = []
    buffer = []
    expected_columns = len(healthy_df.columns)

    for row in corrupted_df.itertuples(index=False, name=None):
        row_data = ','.join(map(str, row))  # Combine row into a string
        split_row = row_data.split(',')  # Split by commas

        if len(split_row) == expected_columns:
            if buffer:
                # Merge buffered data and append
                buffer[-1] += ' ' + split_row[0]  # Fix first column
                fixed_rows.append(buffer + split_row[1:])
                buffer = []
            else:
                fixed_rows.append(split_row)
        elif len(split_row) > expected_columns:
            fixed_rows.append(split_row[:expected_columns])
        else:
            buffer = split_row  # Buffer incomplete row

    # Convert fixed rows to DataFrame
    fixed_df = pd.DataFrame(fixed_rows, columns=healthy_df.columns)
    return fixed_df
```

```python
# Fix the corrupted file
fixed_corrupted_df = fix_and_align_corrupted_file(healthy_df, corrupted_df)
```

```python
# Save the fixed file
fixed_corrupted_df.to_csv('fixed_corrupted_file.csv', index=False)
```
