```python
import pandas as pd

# Load the file
file_path = '../universitiesV4.csv'
data = pd.read_csv(file_path)

# Remove the extra indexing column
data = data.drop(columns=['Unnamed: 0'])

# Clean the 'org_id' column by stripping whitespace and replacing empty strings with NaN
data['org_id'] = data['org_id'].astype(str).str.strip()
data['org_id'].replace('', pd.NA, inplace=True)

# Remove rows where 'org_id' is missing or invalid
cleaned_data = data.dropna(subset=['org_id'])
```

```python
# Save the cleaned data to a new file
output_path = '../universitiesV5.csv'  # Update with your desired output path
cleaned_data.to_csv(output_path, index=False)

print("Cleaned data saved to:", output_path)
```
