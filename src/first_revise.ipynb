```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score

# Assuming 'full_articles' DataFrame is loaded and pre-processed.
# For demonstration, let's create a sample DataFrame if it doesn't exist.

# 2. Define the conversion and validation function
def convert_and_validate_year(year):
    try:
        year_num = int(year)
        if 1300 <= year_num <= 1500:
            year_num += 621
        if 1950 <= year_num <= 2025:
            return year_num
        else:
            return np.nan
    except (ValueError, TypeError):
        return np.nan

# 3. Apply the function to create a new, clean column
full_articles['gregorian_year'] = full_articles['year'].apply(convert_and_validate_year)

# 4. Count invalid entries and get final counts of valid years
invalid_count = full_articles['gregorian_year'].isnull().sum()
yearly_counts = full_articles['gregorian_year'].dropna().value_counts().sort_index()
yearly_counts.index = yearly_counts.index.astype(int)

# 5. Prepare data for plotting and regression
# Using all available data points for a better fit
x_data = yearly_counts[:73].index
y_data = yearly_counts[:73].values

# --- KEY CHANGES FOR NON-LINEAR FIT ---

# 6. Calculate the non-linear trendline (3rd-degree polynomial)
# A polynomial fit creates a curved line. The 'degree' determines its complexity.
degree = 3
coeffs = np.polyfit(x_data, y_data, degree)
trendline_poly = np.poly1d(coeffs)

# Generate points for a smooth curve
x_smooth = np.linspace(x_data.min(), x_data.max(), 500)
y_smooth = trendline_poly(x_smooth)

# Calculate R-squared based on the polynomial fit
r2 = r2_score(y_data, trendline_poly(x_data))

# 7. Create the plot
plt.style.use('seaborn-v0_8-whitegrid')
plt.figure(figsize=(12, 7))

# Bar chart for actual counts
plt.bar(x_data, y_data, label='Actual Articles per Year', color='skyblue', alpha=0.7)

# Plot the smooth, non-linear trendline
plt.plot(x_smooth, y_smooth, color='red', linestyle='--', label=f'Polynomial Trend (R² = {r2:.2f})')

# 8. Add titles, labels, and legend
plt.title('Iranian Article Publication Trend (Non-Linear Fit)', fontsize=16)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Number of Articles', fontsize=12)
plt.ylim(bottom=0)

# Create smarter, non-overlapping x-axis labels
step = max(1, len(x_data) // 10)
plt.xticks(ticks=x_data[::step], rotation=45, ha="right")

plt.legend()
plt.tight_layout()

# 9. Save the plot and print results
#plt.savefig('article_publication_trend_NONLINEAR.png')

print("--- Yearly Article Counts (Sorted by Year) ---")
print(yearly_counts)
print(f"\nNumber of invalid or discarded year entries: {invalid_count}")
print("\n✅ Plot with non-linear trend saved as 'article_publication_trend_NONLINEAR.png'")
```
